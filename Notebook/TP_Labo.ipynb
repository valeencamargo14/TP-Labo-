### 3.1.c) (*continuación*) Coherencia Interna de los Datos
Evaluaremos si las columnas calculadas de cada tabla son consistentes con las columnas de las que se derivan, y si hay puntos que no coinciden con los resultados de los partidos.

**Qué se hace:** Verificamos que `balance` sea igual a `ingresos - gasto_bruto` y que `gasto_neto` sea igual a `gasto_bruto - ingresos` (ambos con una tolerancia de 0.05 por redondeos), que `dif_goles` sea `goles_favor - goles_contra`, que `ganados + empatados + perdidos` sumen los `partidos` jugados, que los puntos coincidan con `3 * ganados + empatados` y que, en cada temporada, la suma de goles a favor sea igual a la de goles en contra.

**Por qué:** Como los archivos CSV son de elaboración propia, conviene comprobar que los números no se contradigan entre sí. Si una columna derivada no coincide con las columnas que la componen, puede haber un error de carga o de transcripción que contamine los análisis posteriores. En particular, `gasto_neto` es la variable central del trabajo, por lo que es necesario confirmar que se corresponde con el gasto bruto y los ingresos.

```python
# Coherencia interna
dif_balance = (df_gastos['ingresos'] - df_gastos['gasto_bruto'] - df_gastos['balance']).abs()
print(f'balance distinto de ingresos - gasto_bruto: {(dif_balance > 0.05).sum()} filas')

dif_neto = (df_gastos['gasto_bruto'] - df_gastos['ingresos'] - df_gastos['gasto_neto']).abs()
print(f'gasto_neto distinto de gasto_bruto - ingresos: {(dif_neto > 0.05).sum()} filas')

print(f"\ndif_goles distinto de goles_favor - goles_contra: "
      f"{(df_posiciones['goles_favor'] - df_posiciones['goles_contra'] != df_posiciones['dif_goles']).sum()} filas")
print(f"ganados + empatados + perdidos distinto de partidos: "
      f"{(df_posiciones['ganados'] + df_posiciones['empatados'] + df_posiciones['perdidos'] != df_posiciones['partidos']).sum()} filas")

esperado = 3 * df_posiciones['ganados'] + df_posiciones['empatados']
print('\nFilas donde puntos != 3*ganados + empatados (sanciones):')
display(df_posiciones[df_posiciones['puntos'] != esperado]
        [['temporada', 'club', 'ganados', 'empatados', 'puntos']].assign(esperado=esperado))

goles = df_posiciones.groupby('temporada')[['goles_favor', 'goles_contra']].sum()
print(f'Temporadas donde la suma de goles a favor y en contra no coincide: {(goles.goles_favor != goles.goles_contra).sum()}')
```

**Resultado:** Todas las comprobaciones dieron 0 filas inconsistentes, salvo una: hay tres filas donde los puntos no coinciden con 3 por victoria más empates, y son sanciones reales de deducción de puntos a los siguientes equipos:

## 3.3. Transformaciones y resumen estadístico

### 3.3.a) Transformaciones de Datos
**Qué se hace:** Aplicamos cuatro transformaciones sobre los datos:
1. **Unión de tablas (`pd.merge()`):** unimos el DataFrame de posiciones con el de gastos usando `temporada` y `club` como claves de cruce, y comprobamos que la cantidad de filas no cambie.
2. **Filtrado de filas:** a partir del dataset completo (`df_completo`, 21 temporadas) creamos `df_estudio`, que conserva *únicamente* las temporadas 2005-06 y 2025-26. `df_completo` sigue disponible para el análisis de la serie completa.
3. **Columna derivada `costo_por_punto`:** generamos una nueva métrica (Gasto Neto / Puntos), en millones de euros por punto.
4. **Columna derivada `grupo_tabla` (`pd.cut()`):** convertimos la posición final (del 1 al 20) en tres categorías: Top 4 (puestos 1 a 4), Media tabla (5 a 17) y Descenso (18 a 20).

**Por qué:**
1. La unión relacional es indispensable porque necesitamos que el rendimiento en la cancha (puntos/posición) conviva en la misma fila con la inversión financiera (gasto neto) para cada club. Contar las filas antes y después es una buena práctica para asegurarnos de que el `merge` no haga desaparecer clubes sin avisar.
2. El filtrado crea el conjunto del contraste central de la pregunta (3.3.b, 3.4.b, 3.4.c y 3.5.e). No se descartan las otras 19 temporadas: siguen en `df_completo` y se usan para medir la relación entre gasto y tabla temporada por temporada y para el largo plazo.
3. El costo por punto relaciona la inversión con el rendimiento: cuánto gastó, en neto, cada club por cada punto. Está en euros nominales, por lo que se interpreta dentro de cada temporada y no se compara directamente entre épocas. Cuando el gasto neto es negativo (el club vendió más de lo que compró), el valor es negativo: indica superávit en traspasos y no eficiencia deportiva.
4. Una posición del 1 al 20 sirve para ordenar a los clubes, pero no para comparar categorías. Agrupamos la tabla en tres zonas: los cuatro primeros, la zona de descenso (los últimos tres) y el resto. El Top 4 corresponde a los cuatro cupos de Champions League que la Premier League tuvo durante casi todo el período; en las últimas temporadas, la UEFA sumó un cupo adicional por desempeño europeo (*European Performance Spot*), del que se benefició el quinto puesto de 2024-25, por lo que ese puesto también pudo acceder a la competencia. Se usa el mismo corte en las 21 temporadas para que los grupos sean comparables; el quinto puesto queda dentro de "Media tabla". Esto permite comparar el gasto entre clubes de arriba, del medio y de abajo en el gráfico 3.4.b.


```python
# 1. Primera transformación: Unión de las tablas (Merge)
df_completo = pd.merge(df_posiciones, df_gastos, on=['temporada', 'club'], how='inner')
print(f"Filas antes del merge: posiciones = {len(df_posiciones)}, gastos = {len(df_gastos)} | después del merge: {len(df_completo)}")

# 2. Segunda transformación: Filtrado temporal
df_estudio = df_completo[df_completo['temporada'].isin(['2005-06', '2025-26'])].copy()

# 3. Tercera transformación: Columna derivada (Costo por Punto)
df_estudio['costo_por_punto'] = round(df_estudio['gasto_neto'] / df_estudio['puntos'], 2)

# 4. Cuarta transformación: Columna derivada (grupo de la tabla según la posición final)
df_estudio['grupo_tabla'] = pd.cut(df_estudio['posicion'], bins=[0, 4, 17, 20],
                                   labels=['Top 4 (1-4)', 'Media tabla (5-17)', 'Descenso (18-20)'])

print(f"Dimensiones del nuevo dataset de estudio: {df_estudio.shape}")

print("\n--- Ejemplo de temporada 2005-06 ---")
display(df_estudio[df_estudio['temporada'] == '2005-06'].head(3))

print("\n--- Ejemplo de temporada 2025-26 ---")
display(df_estudio[df_estudio['temporada'] == '2025-26'].head(3))
```

**Resultado:** el `merge` conserva las 420 filas: cada combinación club-temporada de la tabla de posiciones tiene su par en la tabla de gastos, así que no se pierde ningún dato. El dataset de estudio queda con 40 filas (20 clubes × 2 temporadas) y 19 columnas: las 17 del cruce más las dos columnas nuevas, `costo_por_punto` y `grupo_tabla`.

### 3.3.b) Estadísticos de Resumen Agrupados (.groupby)
**Qué se hace:** Calculamos tres estadísticos agrupados con `groupby()`:
1. Sobre `df_estudio` (2005-06 y 2025-26): el gasto neto (cantidad de clubes, promedio y mediana) por temporada y por grupo de la tabla.
2. Sobre `df_estudio`: el costo por punto (promedio, mediana, mínimo y máximo) por temporada, y la cantidad de clubes con gasto neto negativo (superávit).
3. Sobre `df_completo` (las 21 temporadas): qué tan parecido es el orden de gasto al orden de la tabla, temporada por temporada. Se calcula cuántos de los 4 clubes que más gastaron terminaron Top 4, en qué puesto terminaron en promedio los 5 que más gastaron, en qué puesto de gasto estaba el campeón y cuántos puestos separan, en promedio, el puesto de gasto de cada club de su puesto en la tabla. Se muestran 2005-06 y 2025-26, y la serie completa se usa en la sección 3.5.

**Por qué:**
1. Muestra si los clubes de arriba gastan más que los de abajo y cuánto cambió esa diferencia entre las dos temporadas.
2. El costo por punto relaciona el gasto con el rendimiento: cuánto invirtió, en neto, cada club por cada punto obtenido. Como los montos son nominales, sirve para comparar clubes dentro de una misma temporada pero no es directamente comparable entre épocas. Además, en los clubes con superávit el valor es negativo y no mide eficiencia, por lo que se cuenta aparte cuántos clubes están en esa situación.
3. Mide, con números fáciles de leer, cuánto se parece el ranking de gasto al ranking de la tabla y lo compara con lo esperable por azar (por ejemplo, por azar solo 0.8 de los 4 que más gastan terminarían Top 4). Ranquear dentro de cada temporada evita el problema de la inflación: importa quién gastó más que los demás ese año, no los euros absolutos. Es el estadístico que responde de forma directa a la pregunta sobre la relación entre gasto neto y posición, y por eso se calcula en las 21 temporadas.

```python
# Estadístico 1: gasto neto por temporada y grupo de la tabla
resumen_grupos = (df_estudio.groupby(['temporada', 'grupo_tabla'], observed=True)['gasto_neto']
                  .agg(clubes='count', promedio='mean', mediana='median').round(1))
print("=== GASTO NETO POR GRUPO DE LA TABLA (millones de €) ===")
display(resumen_grupos)

# Estadístico 2: costo por punto por temporada
resumen_costo = df_estudio.groupby('temporada')['costo_por_punto'].agg(['mean', 'median', 'min', 'max']).round(2)
print("\n=== COSTO POR PUNTO POR TEMPORADA (millones de € por punto) ===")
display(resumen_costo)

# Clubes con gasto neto negativo (superávit): su costo por punto es negativo
superavit = (df_estudio['gasto_neto'] < 0).groupby(df_estudio['temporada']).sum()
print("\nClubes con gasto neto negativo (superávit) por temporada:")
print(superavit)
```

**Estadístico 3: ranking de gasto y referencias por azar.**
* **Ranking de gasto (`rank_gasto`).** Dentro de cada temporada se ordenan los 20 clubes por `gasto_neto` de mayor a menor (1 = el que más gastó). Se usa `rank(method='min')`: si dos clubes tuvieran exactamente el mismo gasto neto, ambos recibirían el puesto más bajo del empate (dos clubes empatados en el 3.º lugar tendrían el puesto 3 y el siguiente, el 5). La posición en la tabla (`posicion`) no tiene empates.
* **Qué significa "por azar".** Es lo que se esperaría si el orden de gasto fuera una ordenación al azar de los 20 clubes, sin relación con la tabla:
  - **Aciertos Top 4:** cada uno de los 4 clubes que más gastan tiene 4/20 de probabilidad de terminar entre los 4 primeros, así que el valor esperado es 4 × 4/20 = **0.8**.
  - **Posición promedio de los 5 que más gastan:** al azar, esos 5 clubes serían una muestra cualquiera de los puestos 1 a 20, cuyo promedio es (1 + 20)/2 = **10.5**.
  - **Brecha promedio:** la diferencia absoluta esperada entre dos puestos al azar entre 1 y n es (n² − 1)/(3n) = 399/60 ≈ **6.65**, que redondeamos a **6.7**.

```python
# Estadístico 3: ¿qué tan parecido es el orden de gasto al orden de la tabla? (medidas simples)
# Puesto de cada club según su gasto neto dentro de su temporada (1 = el que más gastó)
df_completo['rank_gasto'] = (df_completo.groupby('temporada')['gasto_neto']
                             .rank(ascending=False, method='min').astype(int))

filas = []
for temporada, d in df_completo.groupby('temporada'):
    top5_gasto = d[d['rank_gasto'] <= 5]
    campeon = d[d['posicion'] == 1].iloc[0]
    filas.append({
        'temporada': temporada,
        'aciertos_top4': int(((d['rank_gasto'] <= 4) & (d['posicion'] <= 4)).sum()),   # de los 4 que más gastaron, cuántos quedaron Top 4
        'pos_prom_top5_gasto': round(top5_gasto['posicion'].mean(), 1),               # posición promedio de los 5 que más gastaron
        'brecha_prom': round((d['rank_gasto'] - d['posicion']).abs().mean(), 1),      # puestos de diferencia, en promedio, entre gasto y tabla
        'campeon': campeon['club'],
        'rank_gasto_campeon': int(campeon['rank_gasto']),                             # puesto de gasto del campeón
    })
medidas = pd.DataFrame(filas).set_index('temporada')

print("=== MEDIDAS SIMPLES DE LAS DOS TEMPORADAS DE LA PREGUNTA ===")
display(medidas.loc[['2005-06', '2025-26']])

# Referencias por azar (n = 20 clubes por temporada)
n = 20
azar_aciertos = 4 * 4 / n               # 0.8
azar_pos_top5 = (n + 1) / 2             # 10.5
azar_brecha = (n**2 - 1) / (3 * n)      # 6.65
print(f"Referencia por azar: {azar_aciertos:.1f} aciertos, posición promedio de los 5 que más gastan = {azar_pos_top5:.1f}, brecha promedio = {azar_brecha:.1f} puestos")
```

**Interpretación:**

**Gasto neto por grupo de la tabla.** En las dos temporadas, cuanto mejor es el grupo, más gasta en neto: Top 4 > Media tabla > Descenso. En 2005-06 el Top 4 gastó en promedio 31.7 millones contra 5.5 de los clubes en zona de descenso; en 2025-26, 168.3 contra 23.8. La brecha en dinero pasó de unos 26 a unos 145 millones. En términos relativos, en cambio, la ventaja de los de arriba no creció: el Top 4 gastó 3.3 veces lo que la media tabla en 2005-06 (31.7 vs. 9.6) y 2.5 veces en 2025-26 (168.3 vs. 68.3).

**Qué tan parecido es el orden de gasto al orden de la tabla.** En 2005-06 cada club terminó, en promedio, a 4.1 puestos de donde lo ubicaba su gasto; en 2025-26, a 5.3 puestos. Si el gasto no tuviera ninguna relación con la tabla, la diferencia sería de unos 6.7 puestos, así que en las dos temporadas el gasto se parece al orden de la tabla, algo más en 2005-06 (que es la temporada más alineada de las 21). En ambas, 2 de los 4 clubes que más gastaron terminaron Top 4 (Chelsea y Liverpool en 2005-06; Arsenal y Manchester City en 2025-26). Es la evidencia central para nuestra pregunta: el vínculo se aflojó un poco, pero no desapareció.

**Costo por punto.** El promedio pasó de 0.23 a 1.44 millones por punto (unas 6 veces) y la mediana, de 0.22 a 2.18 (unas 10 veces). Como los montos son nominales, este aumento refleja sobre todo el encarecimiento del mercado que se ve en el gráfico 3.4.a y no un cambio en la eficiencia de los clubes. El indicador debe leerse con cautela: en los clubes con gasto neto negativo (superávit) el costo por punto es negativo, lo que indica que vendieron más de lo que compraron pero no mide eficiencia deportiva, y esos valores reducen el promedio. Estos clubes pasaron de 1 en 2005-06 a 5 en 2025-26, lo que muestra que hoy hay más clubes que financian su plantel vendiendo jugadores.

### 3.4.b) Gasto neto promedio por grupo de la tabla
**Tipo de gráfico:** barras agrupadas (comparación entre categorías).

**Qué se hace:** Comparamos el gasto neto promedio del Top 4, la media tabla y el descenso, con una barra por temporada en cada grupo. Para esto tomamos el resumen `resumen_grupos` (calculado en 3.3.b), lo reordenamos como matriz con `.pivot()` (filas = temporadas, columnas = grupos) y lo transponemos con `.T` para que cada grupo quede en una fila y cada temporada en una columna, que es la forma que necesita `.plot(kind='bar')`.

**Por qué:** La pregunta es si el dinero se asocia con la posición. Comparar el gasto de los clubes de arriba, del medio y de abajo en las dos temporadas muestra si los equipos exitosos gastan más y cómo cambió esa diferencia. Las barras son el tipo adecuado porque se compara la altura entre categorías.

```python
# 1. Resetamos el índice para tener columnas normales ('temporada', 'grupo_tabla', 'promedio')
df_plano = resumen_grupos.reset_index()

# 2. Armamos la matriz: filas = temporada, columnas = grupo_tabla
matriz_gasto = df_plano.pivot(index='temporada', columns='grupo_tabla', values='promedio')

# 3. Transponemos (.T) para que las filas sean el grupo de la tabla y las columnas la temporada
promedios_t = matriz_gasto.T

# Reordenamos las filas para que mantengan el orden lógico del grupo
orden_grupos = ['Top 4 (1-4)', 'Media tabla (5-17)', 'Descenso (18-20)']
promedios_t = promedios_t.reindex(orden_grupos)

# Graficamos
fig, ax = plt.subplots(figsize=(9, 5.5))
promedios_t.plot(kind='bar', ax=ax, color=[AZUL, NARANJA], width=0.75, edgecolor='white')
for contenedor in ax.containers:
    ax.bar_label(contenedor, fmt='%.1f', padding=3, fontsize=9)

ax.set_title('Gasto neto promedio según el grupo: 2005-06 vs. 2025-26', fontsize=13, fontweight='bold')
ax.set_xlabel('Grupo según la posición final')
ax.set_ylabel('Gasto neto promedio (millones de €)')
ax.tick_params(axis='x', rotation=0)
ax.grid(axis='x', visible=False)
ax.legend(title='Temporada')
plt.tight_layout()
plt.savefig('grafico_3_4b_gasto_por_grupo.png', dpi=300, bbox_inches='tight')
plt.show()
```

**Interpretación:** en las dos temporadas el gasto neto promedio baja a medida que se baja en la tabla: el Top 4 gasta más que la media tabla, y esta más que la zona de descenso. Lo que cambia es la escala: en 2005-06 el Top 4 gastó 31.7 millones en promedio y en 2025-26, 168.3. Pero la ventaja relativa del Top 4 sobre la media tabla no creció (3.3 veces en 2005-06 y 2.5 veces en 2025-26).

El gráfico tiene una limitación: son promedios de pocos clubes (4, 13 y 3), así que un solo club los mueve mucho. Por ejemplo, en 2025-26 el Aston Villa, 4º con solo 25.7 millones, baja el promedio del Top 4. Por eso el gráfico 3.4.c muestra a cada club por separado.

### 3.4.c) Relación entre gasto neto y posición final
**Tipo de gráfico:** diagrama de dispersión (relación entre dos variables).
**Qué se hace:** Para cada una de las dos temporadas dibujamos un punto por club, con el gasto neto en el eje horizontal y la posición final en el vertical (invertido, para que el 1º quede arriba). Escribimos en el título cuántos de los 4 clubes que más gastaron terminaron Top 4 y etiquetamos los cuatro clubes con mayor gasto neto y el de menor. Las zonas sombreadas marcan el Top 4 (verde) y el descenso (rojo).
**Por qué:** Es el gráfico que responde directamente a la pregunta de investigación: muestra la relación club por club y permite ver si los puntos se ordenan de arriba a la izquierda hacia abajo a la derecha, y comparar las dos temporadas. Cada panel tiene su propia escala horizontal porque los montos de 2025-26 son varias veces mayores que los de 2005-06.

```python
fig, axs = plt.subplots(1, 2, figsize=(14, 6.5), sharey=True)
colores = {'2005-06': AZUL, '2025-26': NARANJA}

for ax, temporada in zip(axs, ['2005-06', '2025-26']):
    d = df_estudio[df_estudio['temporada'] == temporada]

    ax.axhspan(0.5, 4.5, color=VERDE, alpha=0.10)                           # zona Top 4
    ax.axhspan(17.5, 20.5, color=ROJO, alpha=0.10)                          # zona de descenso
    ax.axvline(0, color='gray', lw=1)                                       # a la izquierda: vendieron más de lo que compraron
    ax.scatter(d['gasto_neto'], d['posicion'], color=colores[temporada], s=70, alpha=0.9, edgecolor='white', zorder=3)

    # Etiquetamos los 4 clubes con mayor gasto neto y el de menor gasto neto
    destacados = pd.concat([d.nlargest(4, 'gasto_neto'), d.nsmallest(1, 'gasto_neto')])
    for _, fila in destacados.iterrows():
        ax.annotate(nombre_corto(fila['club']), (fila['gasto_neto'], fila['posicion']),
                    textcoords='offset points', xytext=(7, 4), fontsize=9)

    ax.margins(x=0.15)
    aciertos = medidas.loc[temporada, 'aciertos_top4']
    ax.set_title(f"{temporada}: {aciertos} de los 4 que más gastaron terminaron Top 4", fontsize=12, fontweight='bold')
    ax.set_xlabel('Gasto neto (millones de €)')

ax.set_ylabel('Posición final (1 = campeón)')
axs[0].set_ylim(20.8, 0.2)                                                  # eje invertido: el 1º arriba
axs[0].set_yticks([1, 4, 8, 12, 16, 20])
fig.suptitle('Gasto neto vs. posición final de cada club en la Premier League', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_3_4c_dispersion_gasto_posicion.png', dpi=300, bbox_inches='tight')
plt.show()
```

**Interpretación:** en 2005-06 los puntos bajan de forma bastante ordenada de izquierda a derecha: el club que más gastó en neto, Chelsea, fue campeón, y Liverpool, también entre los 4 que más gastaron, terminó 3º. En 2025-26 también 2 de los 4 que más gastaron terminaron Top 4 (Arsenal 1º y Manchester City 2º), pero la nube está más desordenada: Tottenham, cuarto en gasto neto con 182.9 millones, terminó 17º; Liverpool gastó 263.6 millones y terminó 5º; en cambio, el Aston Villa fue 4º con solo 25.7 millones y el Bournemouth terminó 6º vendiendo más de lo que compró (-129.3 millones).

En 2025-26 el gasto neto de la temporada se parece algo menos al orden final de la tabla que en 2005-06 (brecha promedio de 5.3 contra 4.1 puestos), aunque en ambas temporadas la brecha está por debajo de lo esperable por azar (6.7) y 5.3 es un valor habitual dentro de la serie (ver 3.5.a). Hay que leerlo con cuidado: es una asociación y no una causa, y el gasto neto de un año no mide los salarios ni el plantel acumulado de temporadas anteriores.

### 3.5.c) ¿Ganó el que más gastó?

**Tipo de gráfico:** puntos conectados por línea.
**Qué se hace:** Para cada temporada mostramos en qué puesto de gasto neto estaba el campeón (1 = el club que más gastó).
**Por qué:** Es la forma más directa de contar la historia: ¿el título lo gana quien más invierte en el mercado de pases, o hay campeones que ganaron gastando poco en neto?

```python
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(x, medidas['rank_gasto_campeon'], marker='o', color=AZUL, lw=1.5, ls='--', alpha=0.6)
colores_pts = [NARANJA if i in destacadas else AZUL for i in x]
ax.scatter(x, medidas['rank_gasto_campeon'], s=110, color=colores_pts, zorder=3)

# Para controlar colisiones, almacenamos las posiciones de las etiquetas que ya dibujamos
posiciones_dibujadas = []

for i, (t, fila) in enumerate(medidas.iterrows()):
    rank = fila['rank_gasto_campeon']
    nombre = fila['campeon'].replace(' FC', '').replace('Manchester ', 'Man. ')

    # Configuración base
    desplazamiento_y = -22 if rank <= 3 else 18
    desplazamiento_x = 0
    va_val = 'top' if rank <= 3 else 'bottom'

    # Algoritmo simple anti-colisión con etiquetas adyacentes
    for prev_x, prev_y, prev_desp_x, prev_desp_y in posiciones_dibujadas:
        if abs(i - prev_x) <= 1.2 and abs(rank - prev_y) <= 1.5:
            # Si están muy cerca, empujamos la etiqueta actual en dirección opuesta o desfasamos en X
            if desplazamiento_y == prev_desp_y:
                desplazamiento_y = -desplazamiento_y  # Invertir verticalmente
                va_val = 'bottom' if va_val == 'top' else 'top'
            # Agregar desplazamiento en X para evitar solape lateral por la rotación de 30°
            desplazamiento_x = 12 if (i > prev_x) else -12

    # Registrar posición final aproximada de esta etiqueta para el siguiente chequeo
    posiciones_dibujadas.append((i, rank, desplazamiento_x, desplazamiento_y))

    ax.annotate(nombre,
                (i, rank),
                textcoords='offset points',
                xytext=(desplazamiento_x, desplazamiento_y),
                ha='center',
                va=va_val,
                fontsize=8,
                rotation=30,
                zorder=4,
                bbox=dict(boxstyle='round,pad=0.2', fc='white', alpha=0.85, ec='none'))

# Le damos más aire arriba (valores cercanos a 0 en el eje invertido) para que no choque con el título
ax.set_ylim(max(medidas['rank_gasto_campeon']) + 3, -3)    # invertido: 1 (mayor gasto) arriba
ax.set_yticks([1, 5, 10, 15, 20])
ax.set_xticks(x); ax.set_xticklabels(etiquetas, rotation=60, ha='right', fontsize=8)
ax.set_title('Puesto de gasto neto del campeón de cada temporada', fontsize=13, fontweight='bold', pad=25)
ax.set_xlabel('Temporada'); ax.set_ylabel('Puesto en gasto neto del campeón\n(1 = el club que más gastó)')
plt.tight_layout()
plt.savefig('grafico_3_5c_campeon_gasto.png', dpi=300, bbox_inches='tight')
plt.show()
```

**Interpretación:** El campeón fue el mayor gastador neto en 4 de 21 temporadas, estuvo entre los 3 primeros en gasto en 8 y entre los 5 primeros en 10. En las otras 11 ganó sin estar entre los cinco clubes que más gastaron, algo notable en casos como Chelsea 2014-15 (18º en gasto neto), Liverpool 2019-20 (18º), Manchester City 2022-23 (17º) y Liverpool 2024-25 (16º). Hay que leerlo con cuidado: el gasto neto resta lo que se vende, y esos clubes tenían plantel y salarios muy altos construidos en años anteriores. Esto refuerza la cautela de que el gasto neto de una temporada no mide el poder económico total del club.

### 3.5.e) Club por club: puesto de gasto vs. puesto en la tabla

**Tipo de gráfico:** dumbbell (dos puntos unidos por una línea para cada club).
**Qué se hace:** Para 2005-06 y 2025-26, cada club aparece en una fila ordenada por su posición final. El punto azul es su puesto en la tabla y el punto naranja su puesto en gasto neto. Cuanto más larga la línea, mayor la diferencia entre lo que gastó y lo que logró. En el título de cada panel figura la *brecha promedio*: cuántos puestos, en promedio, separan a un club de donde lo ubicaba su gasto.

### 3.5.f) Visión de largo plazo: gasto acumulado de 21 años y posición promedio

**Tipo de gráfico:** barras horizontales.

**Qué se hace:** Para los clubes que jugaron al menos 10 de las 21 temporadas, sumamos todo su gasto neto del período y calculamos su posición promedio. Cada barra es el gasto acumulado; al lado de la barra escribimos la cantidad de temporadas jugadas y la posición promedio del club. Las barras azules son los clubes con posición promedio de 7 o mejor.

**Por qué:** El gasto de una sola temporada no incluye el plantel construido en años anteriores. Sumando varios años, el dinero tiene más chances de mostrar su efecto. La pregunta que responde es: *¿los clubes que más gastaron durante el período son los que mejor terminaron en promedio?* Dos cautelas de lectura: (1) los montos están en euros de cada año (sin ajustar por inflación), por lo que la comparación es más directa entre los clubes que jugaron las 21 temporadas; (2) un club con menos temporadas en la liga acumula menos gasto por ese solo hecho, y su posición promedio refleja solo los años en que estuvo en la Premier League. Por eso se indica la cantidad de temporadas junto a cada barra.

```python
from matplotlib.patches import Patch

largo_plazo = (df_completo.groupby('club')
               .agg(temporadas=('temporada', 'count'),
                    gasto_neto_acum=('gasto_neto', 'sum'),
                    posicion_prom=('posicion', 'mean'))
               .query('temporadas >= 10')
               .sort_values('gasto_neto_acum'))
display(largo_plazo.round(1).sort_values('gasto_neto_acum', ascending=False))

clubes = largo_plazo.index.map(nombre_corto)
colores = [AZUL if p <= 7 else 'lightgray' for p in largo_plazo['posicion_prom']]

fig, ax = plt.subplots(figsize=(11, 8))
ax.barh(clubes, largo_plazo['gasto_neto_acum'], color=colores, edgecolor='white')
for y, (gasto, pos, n) in enumerate(zip(largo_plazo['gasto_neto_acum'], largo_plazo['posicion_prom'],
                                         largo_plazo['temporadas'])):
    ax.text(gasto + 20, y, f"{n} temp. | posición promedio: {pos:.1f}", va='center', fontsize=9)
ax.set_xlim(0, largo_plazo['gasto_neto_acum'].max() * 1.35)
ax.set_title('Gasto neto acumulado 2005-06 a 2025-26 y posición promedio de cada club', fontsize=13, fontweight='bold')
ax.set_xlabel('Gasto neto acumulado (millones de €)')
ax.set_ylabel('Club')
ax.grid(axis='y', visible=False)
ax.legend(handles=[Patch(color=AZUL, label='Posición promedio de 7 o mejor'),
                   Patch(color='lightgray', label='Posición promedio peor que 7')], loc='lower right')
plt.tight_layout()
plt.savefig('grafico_3_5f_largo_plazo.png', dpi=300, bbox_inches='tight')
plt.show()
```

**Interpretación:** Los 6 clubes que más gastaron en neto durante los 21 años (Manchester City, Manchester United, Chelsea, Arsenal, Tottenham y Liverpool) son también los 6 con mejor posición promedio, entre 3.9 y 6.4. Los demás clubes, con menos gasto acumulado, tienen posiciones promedio de 9.6 a 15.8.

Entonces: **el gasto de una temporada se asocia solo de forma moderada con la posición final, pero el gasto sostenido en el tiempo sí distingue a los clubes que terminan arriba.** Los seis clubes de mayor gasto acumulado jugaron las 21 temporadas, de modo que su ventaja no se explica por haber estado más años en la liga. Hay excepciones que lo matizan: Leicester (237 millones, posición promedio 9.9, en 10 temporadas) y Everton (400 millones, 9.6) rindieron más de lo que gastaron, mientras que Newcastle (763 millones, 11.2) y West Ham (703 millones, 12.0) rindieron menos. Dos cautelas: (1) el grupo de seis grandes es el que marca la diferencia, y (2) sigue siendo una asociación y no prueba causalidad: los clubes con más hinchas e ingresos también pueden gastar más.

### 3.5.g) Brecha promedio entre gasto y tabla en las 21 temporadas

**Tipo de gráfico:** líneas (serie temporal).
**Qué se hace:** Graficamos la brecha promedio de cada temporada (cuántos puestos separan, en promedio, el puesto de gasto neto de cada club de su posición final), junto con el valor esperable por azar (6.7) y el promedio de las 21 temporadas. Destacamos 2005-06 y 2025-26.
**Por qué:** La brecha promedio es la medida central de la conclusión. Verla en toda la serie permite comprobar si 2005-06 y 2025-26 son casos típicos o extremos, en lugar de comparar dos temporadas aisladas.

```python
x = np.arange(len(medidas))
etiquetas = list(medidas.index)
destacadas = [etiquetas.index('2005-06'), etiquetas.index('2025-26')]
promedio_brecha = medidas['brecha_prom'].mean()

fig, ax = plt.subplots(figsize=(12, 5.5))
ax.plot(x, medidas['brecha_prom'], marker='o', color=AZUL, lw=2, label='Brecha promedio de la temporada')
ax.scatter(destacadas, medidas['brecha_prom'].iloc[destacadas], s=160, color=NARANJA, zorder=3,
           label='2005-06 y 2025-26')
ax.axhline(azar_brecha, color=ROJO, ls='--', lw=1.5, label=f'Azar ({azar_brecha:.1f})')
ax.axhline(promedio_brecha, color='gray', ls=':', lw=1.5, label=f'Promedio de las 21 temporadas ({promedio_brecha:.1f})')

for i in destacadas:
    ax.annotate(f"{medidas['brecha_prom'].iloc[i]:.1f}", (i, medidas['brecha_prom'].iloc[i]),
                textcoords='offset points', xytext=(0, 12), ha='center', fontsize=10, fontweight='bold')

ax.set_ylim(8.7, 3.4)                                         # eje invertido: arriba = más parecido
ax.set_xticks(x); ax.set_xticklabels(etiquetas, rotation=45, ha='right', fontsize=9)
ax.set_title('Brecha promedio entre el puesto de gasto neto y el puesto en la tabla, por temporada', fontsize=13, fontweight='bold')
ax.set_xlabel('Temporada')
ax.set_ylabel('Brecha promedio (puestos)\nmás arriba = gasto y tabla más parecidos')
ax.legend(loc='lower left', fontsize=9)
plt.tight_layout()
plt.savefig('grafico_3_5g_brecha_por_temporada.png', dpi=300, bbox_inches='tight')
plt.show()
```

**Interpretación:** La brecha fue menor que la esperable por azar en 18 de las 21 temporadas. 2005-06 (4.1) es la temporada más alineada de toda la serie, mientras que 2025-26 (5.3) se ubica cerca del promedio de las 21 temporadas (5.8) y por encima de otras temporadas recientes como 2020-21 (4.7) y 2023-24 (4.6). Esto confirma que el aumento de la brecha entre las dos temporadas de la pregunta se explica en parte porque 2005-06 es un extremo, y no indica un cambio de paradigma.

# **Conclusión**
## Respuesta a la pregunta 3.2.c)

**Respuesta a la pregunta de investigación.** El análisis muestra que el gasto neto en el mercado de pases está asociado de forma positiva y consistente con la posición final en la tabla, pero no la determina. Al contrastar 2005-06 y 2025-26, el vínculo se mantiene en lo esencial: el orden de gasto se parece al orden de la tabla en ambas temporadas, con un leve aflojamiento en la más reciente (brecha promedio de 4.1 puestos en 2005-06 y de 5.3 en 2025-26, frente a 6.7 esperables por azar). Dado que la comparación se apoya en dos temporadas y en datos observacionales, los resultados describen una asociación y no permiten afirmar una relación causal ni un cambio de paradigma.

**Hallazgos principales**

* **Fuerte crecimiento nominal del mercado.** El gasto bruto total de la liga pasó de 505.45 a 4061.22 millones de euros (unas 8 veces) y el gasto neto total, de 268.55 a 1633.14 millones (unas 6 veces). Todos los montos son nominales, es decir, no están ajustados por inflación. La diferencia de gasto neto promedio entre un club del Top 4 y uno en zona de descenso pasó de unos 26 a unos 145 millones de euros. En términos relativos, en cambio, la ventaja del Top 4 sobre la media tabla no creció: fue de 3.3 veces en 2005-06 y de 2.5 veces en 2025-26.

* **2005-06: la mayor alineación entre gasto y tabla de la serie.** En promedio, cada club terminó a 4.1 puestos de donde lo ubicaba su gasto neto, el valor más bajo de las 21 temporadas. El campeón, Chelsea, fue el club con mayor gasto neto de esa temporada.

* **2025-26: un vínculo algo más laxo, pero dentro de lo habitual.** La brecha promedio fue de 5.3 puestos, cercana a la media de las 21 temporadas (5.8) y por debajo del valor esperable por azar (6.7). Hubo discrepancias marcadas: Tottenham fue 4º en gasto neto y terminó 17º, mientras que Aston Villa terminó 4º con un gasto neto de 25.7 millones y Bournemouth terminó 6º con un gasto neto negativo (-129.3 millones). Aun así, el campeón, Arsenal, fue nuevamente el club de mayor gasto neto.

* **Mismo resultado en el Top 4.** En ambas temporadas, 2 de los 4 clubes que más gastaron en neto terminaron entre los 4 primeros. En el conjunto de las 21 temporadas el promedio fue de 1.5 aciertos sobre 4, casi el doble de lo esperable por azar (0.8) pero lejos del máximo posible; nunca se alcanzaron los 4 aciertos y solo una temporada (2014-15) llegó a 3.

* **El gasto de una temporada ayuda, pero no decide el campeonato.** Los cinco clubes que más gastaron terminaron, en promedio, en el puesto 7.6 (contra 10.5 esperable por azar) y mejor ubicados que el azar en 19 de las 21 temporadas. Sin embargo, el campeón fue el club de mayor gasto neto en solo 4 de las 21 temporadas, y en 11 de ellas se ubicó fuera de los cinco primeros en gasto.

* **El gasto sostenido distingue a los clubes de la cima.** Entre los clubes con al menos 10 temporadas en la muestra, los seis de mayor gasto neto acumulado (Manchester City, Manchester United, Chelsea, Arsenal, Tottenham y Liverpool) son también los seis con mejor posición promedio (entre 3.9 y 6.4); el resto se ubica entre 9.6 y 15.8. Hay excepciones que matizan la relación: Everton y Leicester rindieron por encima de lo que gastaron, y Newcastle y West Ham, por debajo.

**Limitaciones metodológicas**

1. **Efecto de los extremos temporales.** La comparación entre 2005-06 y 2025-26 puede exagerar la idea de cambio: 2005-06 es la temporada de mayor alineación de toda la serie, mientras que 2025-26 se ubica dentro de los valores típicos. Otras temporadas, como 2020-21 (4.7) y 2023-24 (4.6), tuvieron brechas menores que la de 2025-26.
2. **Valores nominales.** Los montos están en euros corrientes. Las métricas de ranking (puestos de gasto, aciertos, brecha) mitigan este problema porque comparan a los clubes dentro de cada temporada, pero los promedios por grupo, el costo por punto y los acumulados de largo plazo siguen siendo nominales. En el acumulado, la comparación es más directa entre clubes que disputaron las 21 temporadas.
3. **Variables omitidas.** El análisis se basa solo en transferencias. No incluye la masa salarial, los ingresos de los clubes ni el valor acumulado del plantel, que explican buena parte del rendimiento deportivo.
4. **Asociación y no causalidad.** Los clubes con más hinchas e ingresos pueden gastar más y, a la vez, rendir mejor; el sentido de la relación no puede establecerse con estos datos.
5. **Muestra pequeña y sin contraste estadístico.** La diferencia entre 4.1 y 5.3 puestos surge de 20 clubes por temporada y no se acompañó de una prueba de significatividad, por lo que conviene evitar conclusiones definitivas sobre un cambio en el tiempo.
6. **Calidad y alcance de los datos.** Los montos de Transfermarkt son estimaciones y la composición de la liga cambia cada año por ascensos y descensos. Además, 21 de las 420 observaciones (5 %) tienen ingresos nulos, que en la fuente figuran con un guion: no hay ingresos con monto registrado para ese club y temporada, y no es posible distinguir entre ausencia real de ventas con pago y salidas sin monto informado. En esos casos el gasto neto coincide con el bruto y podría estar sobreestimado. Dos de estas observaciones corresponden a 2005-06 (Bolton y Middlesbrough), una de las dos temporadas centrales del análisis, y ninguna a 2025-26. Por último, el análisis de largo plazo se limita a clubes con al menos 10 temporadas, lo que excluye a quienes pasaron menos tiempo en la liga.

**Posibles extensiones.** Deflactar los montos para comparar euros reales, calcular la correlación de rangos de Spearman por temporada, simular la brecha esperable por azar para medir la incertidumbre, incorporar salarios e ingresos, y estimar un modelo de regresión con datos de panel.
