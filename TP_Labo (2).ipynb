### 3.3 Transformaciones y resumen estadístico
**Qué se hace:** Del dataset `df_gastos` se filtran las temporadas 2025/26 y 2005/06 para, posteriormente, analizar cuántos equipos han tenido un gasto neto mayor o igual a 100 millones. Luego, se suman los gastos de los clubes que han pasado por la Premier League a lo largo de las 21 temporadas para identificar cuáles fueron los que más gastaron.

```python
# Filtrar las temporadas
temp_25_26 = df_gastos[df_gastos["temporada"] == "2025-26"]
temp_05_06 = df_gastos[df_gastos["temporada"] == "2005-06"]
```

```python
# Análisis de la temporada 2025/26
gasto_neto_25_26 = temp_25_26[temp_25_26["gasto_neto"] >= 100]
print(gasto_neto_25_26.shape)
print("Equipos con gasto neto mayor o igual a cien millones (temporada 25/26):", len(gasto_neto_25_26))
```
